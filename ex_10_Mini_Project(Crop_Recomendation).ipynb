```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Sample Crop Recommendation Dataset
data = {
    "N": [90, 85, 88, 40, 42, 45, 70, 75, 72, 25, 30, 28],
    "P": [42, 38, 40, 60, 65, 62, 45, 48, 44, 55, 58, 52],
    "K": [43, 40, 42, 20, 22, 21, 35, 38, 36, 25, 28, 26],
    "temperature": [25, 26, 24, 22, 23, 21, 28, 27, 29, 20, 21, 19],
    "humidity": [80, 82, 78, 65, 68, 66, 70, 72, 68, 60, 62, 58],
    "ph": [6.5, 6.7, 6.4, 6.0, 6.2, 5.9, 6.8, 6.9, 6.7, 5.8, 6.0, 5.7],
    "rainfall": [200, 210, 190, 120, 130, 125, 160, 170, 155, 90, 95, 85],
    "Crop": [
        "Rice", "Rice", "Rice",
        "Wheat", "Wheat", "Wheat",
        "Maize", "Maize", "Maize",
        "Cotton", "Cotton", "Cotton"
    ]
}

df = pd.DataFrame(data)
print("Crop Recommendation Dataset")
print(df)

# Input features
X = df[["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]]

# Target variable
y = df["Crop"]

# Split dataset (test_size=0.33 provides 4 samples in the test set, matching the 4 classes)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=42,
    stratify=y
)

# Create Naive Bayes model
model = GaussianNB()

# Train the model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nActual Crops:")
print(y_test.values)

print("\nPredicted Crops:")
print(y_pred)

print("\nModel Accuracy:", round(accuracy * 100, 2), "%")

# New input for crop recommendation
new_data = pd.DataFrame({
    "N": [88],
    "P": [40],
    "K": [42],
    "temperature": [25],
    "humidity": [80],
    "ph": [6.5],
    "rainfall": [200]
})

# Predict suitable crop
recommendation = model.predict(new_data)

print("\nNew Soil and Weather Conditions:")
print(new_data)

print("\nRecommended Crop:", recommendation[0])
```
